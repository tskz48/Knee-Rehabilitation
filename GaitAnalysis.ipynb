```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import GroupShuffleSplit
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Conv1D, LSTM, Dropout, Layer,ConvLSTM1D,Flatten, GRU
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras import backend as K
from google.colab import drive
import matplotlib.pyplot as plt
import random
drive.mount('/content/drive')


random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)


DATA_PATH = "/content/drive/MyDrive/GaitData.csv"

df = pd.read_csv(DATA_PATH)

print(df)
print(df.dtypes)


feature_cols=['X1', 'X2', 'X3', 'Y4', 'Y5', 'Y6']

SEQ_LENGTH = 25
FORECAST_LENGTH = 5
N_FEATURES=len(feature_cols)




def create_sequences(df_in, seq_length, forecast_length,stride=5):

    data =df_in[feature_cols].values.astype(np.float32)
    X_list, y_list = [], []


    n=len(data)


    for i in range(0, n - seq_length - forecast_length + 1,stride):
            X_list.append(data[i:i + seq_length])
            y_list.append(data[i + seq_length:i + seq_length + forecast_length].reshape(-1))


    return np.array(X_list, dtype=np.float32), np.array(y_list, dtype=np.float32)


test_timesteps=75
trainval_df=df.iloc[:-test_timesteps].copy()
test_df=df.iloc[-test_timesteps:].copy()

val_fraction=0.2
val_start=int(len(trainval_df)*(1-val_fraction))

train_df=trainval_df.iloc[:val_start].copy()
val_df=trainval_df.iloc[val_start:].copy()

print(train_df.shape)
print(val_df.shape)
print(test_df.shape)

X_train, y_train = create_sequences(train_df, SEQ_LENGTH, FORECAST_LENGTH)
X_val,   y_val   = create_sequences(val_df,   SEQ_LENGTH, FORECAST_LENGTH)
X_test,   y_test   = create_sequences(test_df,   SEQ_LENGTH, FORECAST_LENGTH)

print(X_train.shape)
print(X_val.shape)
print(X_test.shape)
print(y_train.shape)
print(y_val.shape)
print(y_test.shape)

X_train_clstm=X_train[...,np.newaxis]
X_val_clstm=X_val[...,np.newaxis]
X_test_clstm=X_test[...,np.newaxis]

print(X_train_clstm.shape)
print(X_val_clstm.shape)
print(X_test_clstm.shape)




class AttentionLayer(tf.keras.layers.Layer):
  def __init__(self, **kwargs):
    super().__init__(**kwargs)

  def build(self,inputs_shape):
    self.W=self.add_weight(shape = (inputs_shape[-1],1), initializer="random_normal",trainable=True, name="att_weight")
    self.b=self.add_weight(shape = (inputs_shape[1],1),initializer='zeros',trainable=True,name="att_bias")

    super().build(inputs_shape)

  def call(self,inputs):
    score =tf.tanh(tf.matmul(inputs,self.W)+self.b)
    weights=tf.nn.softmax(score,axis=1)
    context=tf.reduce_sum(inputs*weights,axis=1)
    return context

def create_model(input_shape, forecast_length):
    inputs = Input(shape=input_shape)

    x = Conv1D(filters=64, kernel_size=3, padding="same", activation="relu")(inputs)
    x = Dropout(0.3)(x)
    x = Conv1D(filters=64, kernel_size=3, padding="same", activation="relu")(x)
    x = Dropout(0.3)(x)
    x = Conv1D(filters=64, kernel_size=3, padding="same", activation="relu")(x)
    x = Dropout(0.3)(x)

    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)

    x = AttentionLayer()(x)
    x = Dropout(0.3)(x)
    outputs=Dense(forecast_length*N_FEATURES)(x)

    model = Model(inputs, outputs)
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.07, momentum=0.9,clipnorm=1.0)
    model.compile(
        optimizer=optimizer,
        loss="mse",
        metrics=[tf.keras.metrics.MeanAbsoluteError(),
                 tf.keras.metrics.RootMeanSquaredError()]
    )
    return model


def create_conv_lstm_model(input_shape,forecast_length):
  inputs=Input(shape=input_shape)
  x=ConvLSTM1D(filters=32,kernel_size=3,padding='same',return_sequences=True,activation='tanh',recurrent_activation='sigmoid')(inputs)
  x=Dropout(0.3)(x)
  x=ConvLSTM1D(filters=32,kernel_size=3,padding='same',return_sequences=False,activation='tanh',recurrent_activation='sigmoid')(x)
  x=Dropout(0.3)(x)
  x=Flatten()(x)
  outputs=Dense(forecast_length*N_FEATURES)(x)
  model=Model(inputs,outputs)
  optimizer=tf.keras.optimizers.SGD(learning_rate=0.07, momentum=0.9,clipnorm=1.0)
  model.compile(optimizer=optimizer,loss='mse',metrics=[tf.keras.metrics.MeanAbsoluteError(),tf.keras.metrics.RootMeanSquaredError()])

  return model

def create_gru_model(input_shape,forecast_length):
  inputs=Input(shape=input_shape)
  x=GRU(128,return_sequences=True)(inputs)
  x=Dropout(0.3)(x)
  x=GRU(128, return_sequences=False)(x)
  x=Dropout(0.3)(x)
  outputs=Dense(forecast_length*N_FEATURES)(x)
  model=Model(inputs,outputs)
  optimizer=tf.keras.optimizers.SGD(learning_rate=0.07,momentum=0.9,clipnorm=1.0)
  model.compile(optimizer=optimizer,loss='mse',metrics=[tf.keras.metrics.MeanAbsoluteError(),tf.keras.metrics.RootMeanSquaredError()])

  return model




def create_lstm_attention_model(input_shape,forecast_length):
    inputs=Input(shape=input_shape)

    x = LSTM(256, return_sequences=True)(inputs)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)
    x = LSTM(256, return_sequences=True)(x)
    x = Dropout(0.3)(x)

    x = AttentionLayer()(x)
    x = Dropout(0.3)(x)
    outputs=Dense(forecast_length*N_FEATURES)(x)

    model = Model(inputs, outputs)
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.07, momentum=0.9,clipnorm=1.0)
    model.compile(
        optimizer=optimizer,
        loss="mse",
        metrics=[tf.keras.metrics.MeanAbsoluteError(),
                 tf.keras.metrics.RootMeanSquaredError()]
    )

    return model











tf.keras.backend.clear_session()

model = create_model((SEQ_LENGTH, X_train.shape[-1]), FORECAST_LENGTH)
model.summary()
early_stopping = EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor="val_loss", factor=0.2, patience=5, min_lr=1e-5)
print('\n Training CNN-LSTM-Attention model... \n')
history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=64,
    validation_data=(X_val, y_val),
    callbacks=[early_stopping, reduce_lr],
)

y_pred = model.predict(X_test, verbose=0)
print(y_pred.shape)
y_test_reshaped = y_test.reshape(-1, FORECAST_LENGTH, N_FEATURES)
print(y_test.shape)
y_pred_reshaped = y_pred.reshape(-1, FORECAST_LENGTH, N_FEATURES)
print(y_pred_reshaped.shape)
results = []

for j, var_name in enumerate(feature_cols):
    true_var = y_test_reshaped[:, :, j].reshape(-1)
    pred_var = y_pred_reshaped[:, :, j].reshape(-1)

    mae_var = np.mean(np.abs(pred_var - true_var))
    rmse_var = np.sqrt(np.mean((pred_var - true_var) ** 2))

    value_range=np.max(true_var)-np.min(true_var)
    nrmse_var=(rmse_var/value_range) *100 if value_range !=0 else np.nan
    cc_var = np.corrcoef(true_var,pred_var)[0,1]

    results.append({
        "Variable": var_name,
        "MAE": mae_var,
        "RMSE": rmse_var,
        "NRMSE_%":nrmse_var,
        "CC":cc_var
    })


conv_lstm_model=create_conv_lstm_model((SEQ_LENGTH,N_FEATURES,1),FORECAST_LENGTH)

early_stopping_clstm=EarlyStopping(monitor="val_loss",patience=10,restore_best_weights=True)
reduce_lr_clstm=ReduceLROnPlateau(monitor="val_loss",factor=0.2,patience=5,min_lr=1e-5)
print('\n Training ConvLSTM model... \n')
history_conv_lstm=conv_lstm_model.fit(X_train_clstm,y_train,epochs=100,batch_size=64,validation_data=(X_val_clstm,y_val),callbacks=[early_stopping_clstm,reduce_lr_clstm])

y_pred_convlstm=conv_lstm_model.predict(X_test_clstm,verbose=0)
print(y_pred_convlstm.shape)
y_pred_convlstm_reshaped=y_pred_convlstm.reshape(-1,FORECAST_LENGTH,N_FEATURES)
print(y_pred_convlstm_reshaped.shape)
results_convlstm=[]

for j, var_name in enumerate(feature_cols):
    true_var = y_test_reshaped[:, :, j].reshape(-1)
    pred_var = y_pred_convlstm_reshaped[:, :, j].reshape(-1)

    mae_var = np.mean(np.abs(pred_var - true_var))
    rmse_var = np.sqrt(np.mean((pred_var - true_var) ** 2))

    value_range=np.max(true_var)-np.min(true_var)
    nrmse_var=(rmse_var/value_range) *100 if value_range !=0 else np.nan
    cc_var = np.corrcoef(true_var,pred_var)[0,1]

    results_convlstm.append({
        "Variable": var_name,
        "MAE": mae_var,
        "RMSE": rmse_var,
        "NRMSE_%":nrmse_var,
        "CC":cc_var
    })


gru_model=create_gru_model((SEQ_LENGTH,N_FEATURES),FORECAST_LENGTH)

early_stopping_gru=EarlyStopping(monitor="val_loss",patience=10,restore_best_weights=True)
reduce_lr_gru=ReduceLROnPlateau(monitor="val_loss",factor=0.2,patience=5,min_lr=1e-5)
print('\n Training GRU model... \n')
history_GRU=gru_model.fit(X_train,y_train,epochs=100,batch_size=64,validation_data=(X_val,y_val),callbacks=[early_stopping_gru,reduce_lr_gru])

y_pred_gru=gru_model.predict(X_test,verbose=0)
print(y_pred_gru.shape)
y_pred_gru_reshaped=y_pred_gru.reshape(-1,FORECAST_LENGTH,N_FEATURES)
print(y_pred_gru_reshaped.shape)
results_gru=[]

for j, var_name in enumerate(feature_cols):
    true_var = y_test_reshaped[:, :, j].reshape(-1)
    pred_var = y_pred_gru_reshaped[:, :, j].reshape(-1)

    mae_var = np.mean(np.abs(pred_var - true_var))
    rmse_var = np.sqrt(np.mean((pred_var - true_var) ** 2))

    value_range=np.max(true_var)-np.min(true_var)
    nrmse_var=(rmse_var/value_range) *100 if value_range !=0 else np.nan
    cc_var = np.corrcoef(true_var,pred_var)[0,1]

    results_gru.append({
        "Variable": var_name,
        "MAE": mae_var,
        "RMSE": rmse_var,
        "NRMSE_%":nrmse_var,
        "CC":cc_var
    })




lstm_attention_model=create_lstm_attention_model((SEQ_LENGTH,N_FEATURES),FORECAST_LENGTH)

early_stopping_lstm_att=EarlyStopping(monitor="val_loss",patience=10,restore_best_weights=True)
reduce_lr_lstm_att=ReduceLROnPlateau(monitor="val_loss",factor=0.2,patience=5,min_lr=1e-5)
print('\n Training LSTM-Attention model... \n')
history_lstm_att=lstm_attention_model.fit(X_train,y_train,epochs=100,batch_size=64,validation_data=(X_val,y_val),callbacks=[early_stopping_lstm_att,reduce_lr_lstm_att])

y_pred_lstm_att=lstm_attention_model.predict(X_test,verbose=0)
print(y_pred_lstm_att.shape)
y_pred_lstm_att_reshaped=y_pred_lstm_att.reshape(-1,FORECAST_LENGTH,N_FEATURES)
print(y_pred_lstm_att_reshaped.shape)
results_lstm_att=[]

for j, var_name in enumerate(feature_cols):
    true_var = y_test_reshaped[:, :, j].reshape(-1)
    pred_var = y_pred_lstm_att_reshaped[:, :, j].reshape(-1)

    mae_var = np.mean(np.abs(pred_var - true_var))
    rmse_var = np.sqrt(np.mean((pred_var - true_var) ** 2))

    value_range=np.max(true_var)-np.min(true_var)
    nrmse_var=(rmse_var/value_range) *100 if value_range !=0 else np.nan
    cc_var = np.corrcoef(true_var,pred_var)[0,1]

    results_lstm_att.append({
        "Variable": var_name,
        "MAE": mae_var,
        "RMSE": rmse_var,
        "NRMSE_%":nrmse_var,
        "CC":cc_var
    })




seed=X_test[0]
print(seed.shape)
print(X_test.shape)

results_df = pd.DataFrame(results)
results_convlstm_df=pd.DataFrame(results_convlstm)
results_gru_df=pd.DataFrame(results_gru)
results_lstm_att_df=pd.DataFrame(results_lstm_att)
print("\nCNN-LSTM-Attention Per-variable test metrics:")
print(results_df)

print('\nConvLSTM Per-variable test metrics:')
print(results_convlstm_df)

print('\nGRU Per-variable test metrics:')
print(results_gru_df)

print('\nLSTM-Attention Per-variable test metrics:')
print(results_lstm_att_df)




model_names =['CNN-LSTM-Attention', 'ConvLSTM', 'GRU','LSTM-Attention']


plot_order=['X1', 'Y4', 'X2', 'Y5', 'X3', 'Y6']
y_label_map = {
    "X1": 'Angular Velocity',
     "X2": 'Angular Velocity',
      "X3": 'Angular Velocity',
       "Y4": 'Linear Acceleration',
        "Y5": 'Linear Acceleration',
         "Y6": 'Linear Acceleration',
}
fig,axes = plt.subplots(3,2,figsize=(14,12))
axes=axes.flatten()
for j, var_name in enumerate(plot_order):
  idx=feature_cols.index(var_name)
  true_var = y_test_reshaped[:,:,idx].reshape(-1)
  pred_cnn_lstm_att = y_pred_reshaped[:,:,idx].reshape(-1)
  pred_convlstm = y_pred_convlstm_reshaped[:,:,idx].reshape(-1)
  pred_gru=y_pred_gru_reshaped[:,:,idx].reshape(-1)
  pred_lstm_att=y_pred_lstm_att_reshaped[:,:,idx].reshape(-1)
  ax=axes[j]

  ax.plot(true_var,label='True', color='black')
  ax.plot(pred_cnn_lstm_att,label='CNN-LSTM-Attention',color='blue',linestyle='--',linewidth=1.8)
  ax.plot(pred_convlstm,label='convlstm',color='red',linestyle=':',linewidth=1.8)
  ax.plot(pred_gru,label='GRU',color='green',linestyle='-.',linewidth=1.8)
  ax.plot(pred_lstm_att,label='LSTM-ATT', color='purple',linestyle='-.', linewidth=1.8)
  ax.set_xlabel('Timestep')
  ax.set_ylabel(y_label_map[var_name])
  ax.set_title(var_name)
  ax.grid(True)
handles,labels=axes[0].get_legend_handles_labels()
fig.legend(handles,labels,loc='upper center', bbox_to_anchor=(0.5,0.94), ncol=5)
fig.suptitle('True vs Predicted Trajectories',y=0.99)
plt.tight_layout(rect=[0,0,1,0.90])
plt.show()




x=np.arange(len(model_names))
fig,axes=plt.subplots(3,2,figsize=(14,12))
axes=axes.flatten()
for j, var_name in enumerate(plot_order):
  ax=axes[j]
  ax2=ax.twinx()
  rmse_vals=[results_df.loc[results_df['Variable']==var_name,'RMSE'].values[0],
  results_convlstm_df.loc[results_convlstm_df['Variable']==var_name,'RMSE'].values[0],
             results_gru_df.loc[results_gru_df['Variable']==var_name, 'RMSE'].values[0],
             results_lstm_att_df.loc[results_lstm_att_df['Variable']==var_name,'RMSE'].values[0]]

  cc_vals=[results_df.loc[results_df['Variable']==var_name,'CC'].values[0],
  results_convlstm_df.loc[results_convlstm_df['Variable']==var_name,'CC'].values[0],
           results_gru_df.loc[results_gru_df['Variable']==var_name, 'CC'].values[0],
           results_lstm_att_df.loc[results_lstm_att_df['Variable']==var_name, 'CC'].values[0]]

  ax.plot(x,rmse_vals,marker='s', color='red',linestyle='-',linewidth=2,label='RMSE')
  ax2.plot(x,cc_vals,marker='o', color='black', linestyle='--',linewidth=2,label='CC')
  ax.set_xticks(x)
  ax.set_xticklabels(model_names,rotation=20)
  ax.set_title(var_name)
  ax.set_ylabel('RMSE',color='red')
  ax2.set_ylabel('CC',color='black')
  ax.tick_params(axis='y',colors='red')
  ax2.tick_params(axis='y',colors='black')
  ax.grid(True)
fig.suptitle('Model comparison RMSE and CC',y=0.98)
plt.tight_layout(rect=[0,0,1,0.96])
plt.show()



fig,axes=plt.subplots(3,2,figsize=(14,12))
axes=axes.flatten()
for j, var_name in enumerate(plot_order):
  ax=axes[j]
  ax2=ax.twinx()
  mae_vals=[results_df.loc[results_df['Variable']==var_name,'MAE'].values[0],
  results_convlstm_df.loc[results_convlstm_df['Variable']==var_name,'MAE'].values[0],
             results_gru_df.loc[results_gru_df['Variable']==var_name,'MAE'].values[0],
             results_lstm_att_df.loc[results_lstm_att_df['Variable']==var_name,'MAE'].values[0]]

  nrmse_vals=[results_df.loc[results_df['Variable']==var_name,'NRMSE_%'].values[0],
              results_convlstm_df.loc[results_convlstm_df['Variable']==var_name,'NRMSE_%'].values[0],
              results_gru_df.loc[results_gru_df['Variable']==var_name,'NRMSE_%'].values[0],
              results_lstm_att_df.loc[results_lstm_att_df['Variable']==var_name,'NRMSE_%'].values[0]]

  ax.plot(x,mae_vals,marker='s',color='green',linestyle='-',linewidth=2,label='MAE')
  ax2.plot(x,nrmse_vals,marker='o',color='blue',linestyle='--',linewidth=2,label='NRMSE')
  ax.set_xticks(x)
  ax.set_xticklabels(model_names,rotation=20)
  ax.set_title(var_name)
  ax.set_ylabel('MAE',color='green')
  ax2.set_ylabel('NRMSE (%)',color='blue')
  ax.tick_params(axis='y',colors='green')
  ax2.tick_params(axis='y',colors='blue')
  ax.grid(True)
fig.suptitle('Model comparison MAE and NRMSE', y=0.98)
plt.tight_layout(rect=[0,0,1,0.96])
plt.show()

```
